This notebook investigates:
1. Acoustic distributions of mel-spectrograms and chroma across genres.
2. Music structure graph topology (node count, degree distribution, edge density).

```python
import glob
import torch
import numpy as np
import matplotlib.pyplot as plt
from src.audio_features import AudioFeatureExtractor
from src.graph_builder import MusicStructureGraphBuilder

extractor = AudioFeatureExtractor(sr=22050)
builder = MusicStructureGraphBuilder(similarity_threshold=0.65)
print("EDA components loaded.")
```

```python
# Analyze preprocessed graphs in data/processed/graphs/
graph_files = sorted(glob.glob("data/processed/graphs/*.pt"))
print(f"Total preprocessed graphs analyzed: {len(graph_files)}")

node_counts = []
edge_counts = []
for gf in graph_files:
    g = torch.load(gf, weights_only=False)
    node_counts.append(g.num_nodes)
    edge_counts.append(g.edge_index.shape[1])

print(f"Average nodes per graph: {np.mean(node_counts):.1f} (std: {np.std(node_counts):.1f})")
print(f"Average edges per graph: {np.mean(edge_counts):.1f} (std: {np.std(edge_counts):.1f})")
```
